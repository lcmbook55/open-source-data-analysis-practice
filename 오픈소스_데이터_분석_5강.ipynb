## 5-2 JSON 형식 저장

```python
### 라이브러리 임포트
import json
import pandas as pd

data = {
    "이름": "홍길동",
    "나이": 25,
    "거주지": "서울",
    "관심사": ["프로그래밍", "데이터 분석", "여행"]
}
## 참고사항
#### indent →  들여쓰기
#### DataFrame → 행과 열로 이루어진 표
#### orient → JSON으로 변환할 때 데이터를 어떻게 배치할지 결정
#### records → 각 행을 하나의 객체로 저장

## json.dump를 이용한 저장 - json 원본 구조 그대로 저장됨(ensure_ascii사용)
with open('output.json', mode='w', encoding='utf-8') as f:
    json.dump(data, f, indent=4, ensure_ascii=False)

## DataFrame을 이용한 저장 - 행 단위로 저장됨(force_ascii사용)
### orient='records'는 DataFrame의 데이터를 한 행(row)씩 묶어서 JSON 객체로 저장하라는 뜻
### orient의 값에 따른 차이
    #### orient='records' - 행을 기준으로 묶기
    #### orient='columns' - 열을 기준으로 묶기
    #### orient='index' - 인덱스를 기준으로 묶기

df = pd.DataFrame(data)
df.to_json('output_df.json', orient='records', indent=4, force_ascii=False)
```
